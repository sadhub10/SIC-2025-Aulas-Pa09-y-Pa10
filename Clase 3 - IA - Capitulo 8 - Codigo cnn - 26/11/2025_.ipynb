# 🚀 PRÁCTICA 1: Red Neuronal Convolucional (CNN) - Flujo de Trabajo

**Objetivo:** Implementar una CNN para clasificación de imágenes, separando el proceso en pasos claros.

## ⚙️ Paso 1: Importar Bibliotecas y Cargar Datos

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, Flatten, MaxPooling2D
import numpy as np

# Cargar datos de MNIST (dígitos escritos a mano)
(X_train, Y_train), (X_test, Y_test) = tf.keras.datasets.mnist.load_data()
```

## 🧹 Paso 2: Preprocesamiento y Normalización

Normalizar la intensidad de los píxeles (de 0-255 a 0-1) y ajustar la forma de los datos para la capa Conv2D (añadir el canal de color/escala de grises).

```python
# Normalizar a [0, 1]
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0

# Expandir dimensiones: de (60000, 28, 28) a (60000, 28, 28, 1)
# El '1' es el canal de color (escala de grises)
X_train = np.expand_dims(X_train, -1)
X_test = np.expand_dims(X_test, -1)

Y_train = Y_train.astype('int32')
Y_test = Y_test.astype('int32')

print(f"Forma de datos de entrenamiento después del preprocesamiento: {X_train.shape}")
```

## 🏗️ Paso 3: Definición de la Arquitectura CNN

Se apilan las capas **Convolucionales**, de **Pooling** y, finalmente, la capa **Densa** de salida.

```python
modelo_cnn_paso = Sequential([
    # 1. Capa Convolucional (32 filtros de 3x3)
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)),

    # 2. Capa de Pooling (reduce la dimensionalidad a la mitad)
    MaxPooling2D(pool_size=(2, 2)),

    # 3. Aplanamiento (necesario antes de la capa Densa)
    Flatten(),

    # 4. Capa Densa de Salida (10 neuronas para 10 dígitos)
    Dense(10, activation='softmax')
])

modelo_cnn_paso.summary()
```

## ⚙️ Paso 4: Compilación del Modelo

Definimos el **Optimizador** (Adam), la **Función de Pérdida** (`loss`) y las **Métricas** (`metrics`) que se usarán.

```python
modelo_cnn_paso.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Modelo compilado y listo para entrenar.")
```

## 🏃‍♂️ Paso 5: Entrenamiento y Evaluación

Aquí la red aprende ajustando sus pesos a través de las épocas.

```python
EPOCHS = 3 # Usamos 3 épocas para un entrenamiento rápido

print("--- Iniciando Entrenamiento ---")
modelo_cnn_paso.fit(X_train, Y_train, epochs=EPOCHS, validation_data=(X_test, Y_test), verbose=1)

# Evaluar rendimiento final en el conjunto de prueba
loss, accuracy = modelo_cnn_paso.evaluate(X_test, Y_test, verbose=0)
print(f"\nPrecisión en datos de prueba: {accuracy*100:.2f}%")
```
