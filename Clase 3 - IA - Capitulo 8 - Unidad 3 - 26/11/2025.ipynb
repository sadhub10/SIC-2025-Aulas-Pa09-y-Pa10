**Objetivo:** Implementar y comparar las tres arquitecturas principales de redes neuronales (MLP, CNN, RNN/LSTM) usando TensorFlow y Keras en casos de uso prácticos.

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, Flatten, MaxPooling2D, LSTM, SimpleRNN
import numpy as np
import matplotlib.pyplot as plt

print("Bibliotecas importadas. ¡Comenzamos la práctica!")
```

## I. Modelo Base: Perceptrón Multicapa (MLP) (Min. 0-30)

### Teoría: Limitaciones del MLP
El **MLP (Multi-Layer Perceptron)** es ideal para datos tabulares. Falla con imágenes o secuencias porque trata cada entrada de datos (píxel, palabra) de forma **independiente**, sin considerar su relación espacial o temporal.

```python
# Nota: Asumimos que tf, np y Sequential ya fueron importados en celdas anteriores.

### Código 1: MLP para Clasificación Básica (Datos Tabulares)

# --- 1. Creación y Simulación de Datos de Ejemplo ---

# Creamos 100 muestras (filas) con 4 características (columnas)
# Esto simula un dataset tabular, como el Iris.
X_mlp = np.random.rand(100, 4)

# Creamos las 100 etiquetas de salida (Y). Los valores son enteros aleatorios (0, 1 o 2).
# Esto representa las 3 clases que el modelo debe aprender a clasificar.
Y_mlp = np.random.randint(0, 3, size=(100, 1))


# --- 2. Definición de la Arquitectura del Modelo MLP ---

# Instanciamos el modelo como Sequential, que apila las capas linealmente.
modelo_mlp = Sequential([
    # Capa Oculta 1 (Entrada): 16 neuronas.
    # 'relu' es la función de activación más común para capas ocultas.
    # 'input_shape=(4,)' indica que el modelo espera 4 características de entrada.
    Dense(units=16, activation='relu', input_shape=(4,)),

    # Capa Oculta 2: 8 neuronas.
    Dense(units=8, activation='relu'),

    # Capa de Salida: 3 neuronas (una por cada clase: 0, 1, 2).
    # 'softmax' transforma las salidas en probabilidades que suman 1.
    Dense(units=3, activation='softmax')
])


# --- 3. Compilación del Modelo ---

modelo_mlp.compile(
    # Optimizador: 'adam' es una versión avanzada del Descenso de Gradiente (SGD)
    # que ajusta la tasa de aprendizaje automáticamente.
    optimizer='adam',

    # Función de Pérdida (Loss): Mide el error.
    # 'sparse_categorical_crossentropy' se usa para clasificación multiclase con etiquetas enteras.
    loss='sparse_categorical_crossentropy',

    # Métrica: 'accuracy' (precisión) es la métrica que monitoreamos para medir el rendimiento.
    metrics=['accuracy']
)


# --- 4. Entrenamiento del Modelo ---

print("--- Entrenando MLP (10 Épocas) ---")

# .fit() inicia el proceso de entrenamiento (forward pass, loss, backpropagation).
# 'epochs=10' indica que el modelo pasará 10 veces por todo el dataset.
# 'verbose=0' desactiva la impresión detallada de cada época para un código más limpio.
history_mlp = modelo_mlp.fit(X_mlp, Y_mlp, epochs=10, verbose=0)

# Imprimimos el resultado de la precisión obtenida en la última época.
print(f"MLP Precisión final: {history_mlp.history['accuracy'][-1]:.4f}")
```

## II. Visión por Computadora: Redes Convolucionales (CNN) (Min. 30-75)

### Analogía del Pintor
El **filtro convolucional (kernel)** es como un pincel especializado que se mueve sobre la imagen para detectar un patrón específico (ej. una línea vertical). Al apilar muchos de estos "pinceles", la CNN aprende a ver el mundo.

### Teoría: Capas Clave
1.  **Conv2D:** Aplica la convolución y extrae características.
2.  **MaxPooling2D:** Reduce la dimensionalidad, haciendo que el modelo sea más robusto a pequeños desplazamientos.

```python
### Código 2: Modelo CNN Funcional (Clasificación de Imágenes)

# 1. Cargar Datos de Imágenes (MNIST - dígitos)
(X_train, Y_train), (X_test, Y_test) = tf.keras.datasets.mnist.load_data()

# Normalizar y dar forma para CNN (28x28x1)
X_train = X_train.astype('float32') / 255.0
X_train = np.expand_dims(X_train, -1)
Y_train = Y_train.astype('int32')

# 2. Definición de la Arquitectura CNN
modelo_cnn = Sequential([
    Conv2D(32, kernel_size=(3, 3), activation='relu', input_shape=(28, 28, 1)), # Capa Convolucional
    MaxPooling2D(pool_size=(2, 2)),                                           # Capa de Pooling
    Flatten(),                                                                # Capa de Aplanamiento
    Dense(10, activation='softmax')                                           # Capa de Salida (10 clases)
])

modelo_cnn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 3. Entrenamiento (Solo 3 épocas para ser rápido)
print("--- Entrenando CNN (MNIST) ---")
history_cnn = modelo_cnn.fit(X_train, Y_train, epochs=3, validation_split=0.1, verbose=1)
print("CNN Entrenada.")
```

## III. Análisis de Secuencias: RNN/LSTM (Min. 75-120)

### Analogía del Lector (Memoria)
Un **RNN/LSTM** es como un lector que, para entender la palabra *'banco'* en una frase, necesita recordar si la frase hablaba de dinero o de un río. Mantiene un **estado interno** que es su memoria.

### Teoría: El Poder del LSTM
**RNN Simple** sufre de **memoria corta** (vanishing gradient). **LSTM (Long Short-Term Memory)** lo soluciona con **compuertas (gates)** que controlan cuándo guardar, olvidar o usar información.

```python
### Código 3: Modelo LSTM para Predicción de Secuencias

# 1. Datos de Secuencia Artificial (Predecir el siguiente número: 1->2, 2->3, etc.)
time_steps = 10
X_rnn = np.array(range(time_steps)) / time_steps
Y_rnn = np.array(range(1, time_steps + 1)) / time_steps

# Las capas RNN/LSTM esperan una forma [muestras, pasos_de_tiempo, características]
X_rnn = X_rnn.reshape(time_steps, 1, 1)
Y_rnn = Y_rnn.reshape(time_steps, 1)

# 2. Definición del Modelo LSTM
modelo_lstm = Sequential([
    # Una capa LSTM de 50 unidades, esperando 1 paso de tiempo y 1 característica
    LSTM(50, activation='relu', input_shape=(1, 1)),
    Dense(1) # Salida (predice el siguiente valor)
])

modelo_lstm.compile(optimizer='adam', loss='mse')

# 3. Entrenamiento
print("--- Entrenando LSTM (Series de Tiempo) ---")
modelo_lstm.fit(X_rnn, Y_rnn, epochs=200, verbose=0)
print("LSTM Entrenada.")

# 4. Predicción de un valor futuro
input_val = np.array([0.7]).reshape((1, 1, 1))
pred_val = modelo_lstm.predict(input_val, verbose=0)[0, 0]
print(f"\nPredicción para 0.7: {pred_val:.4f} (El valor esperado es 0.8)")
```
