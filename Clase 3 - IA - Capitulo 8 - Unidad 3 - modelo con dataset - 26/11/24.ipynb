## 🧹 Paso 2: Preprocesamiento de Datos (Codificación y Separación)

Las redes neuronales solo entienden números. Codificaremos las clases (Setosa, Versicolor, Virginica) a 0, 1 y 2.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import numpy as np

# X: Características (columnas 1-4); Y: Etiquetas (columna 'Species')
X = df.iloc[:, 1:5].values
Y_labels = df.iloc[:, 5].values

# Codificación de etiquetas (Setosa->0, Versicolor->1, Virginica->2)
encoder = LabelEncoder()
Y = encoder.fit_transform(Y_labels)

# Separar datos en entrenamiento y prueba
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
```

## 🏗️ Paso 3: 🏗️ Paso 3: Definición de la Arquitectura MLP (Capas Densa)
Este paso define la estructura de la red neuronal.



```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Estas variables se obtuvieron del Paso 2 (Preprocesamiento del archivo Iris.csv)
# Asumiendo que X_train y Y_train están definidos.
# Si X_train NO está definido, esta celda fallará.
num_features = X_train.shape[1] # 4 características
num_classes = 3 # 3 especies

modelo_iris = Sequential([
    # Capa Oculta 1: 10 neuronas
    Dense(units=10, activation='relu', input_shape=(num_features,)),

    # Capa Oculta 2: 10 neuronas
    Dense(units=10, activation='relu'),

    # Capa de Salida: 3 neuronas con softmax
    Dense(units=num_classes, activation='softmax')
])

print("Arquitectura del Perceptrón Multicapa (MLP):")
modelo_iris.summary()
```

## ⚙️ Paso 4: Compilación del Modelo
Este paso configura el optimizador ( adam), la función de pérdida ( sparse_categorical_crossentropy) y la métrica ( accuracy).

```python
modelo_iris.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Modelo compilado exitosamente.")
```

## 🏃‍♂️ Paso 5: Entrenamiento y Vista Detallada de las Épocas
Aquí se ejecuta el entrenamiento. El parámetro verbose=1es clave para ver el avance de la pérdida y la precisión en cada una de las 50 épocas.

```python
EPOCHS = 80
print(f"--- Entrenando Modelo Iris durante {EPOCHS} épocas ---\n")

# verbose=1 Muestra el progreso de cada época
# Se añade validation_data para evaluar el rendimiento en el conjunto de prueba durante el entrenamiento
modelo_iris.fit(X_train, Y_train, epochs=EPOCHS, verbose=1, validation_data=(X_test, Y_test))
```

## ✔️ Paso 6: Evaluación Final
Finalmente, se evalúa el rendimiento del modelo con el conjunto de datos de prueba ( X_test).

```python
# Evaluación en el conjunto de prueba
loss, accuracy = modelo_iris.evaluate(X_test, Y_test, verbose=0)

print(f"\n--- Resultado Final de Evaluación ---")
print(f"Pérdida (Loss) en datos de prueba: {loss:.4f}")
print(f"Precisión (Accuracy) en datos de prueba: {accuracy*100:.2f}%")
```

## 🚀 Paso 7: Predicciones y Visualización de Confianza

```python
import numpy as np
import pandas as pd

# Seleccionar algunos ejemplos del conjunto de prueba para predecir
# Puedes cambiar los índices para probar diferentes muestras
sample_indices = [0, 5, 10, 15] # Seleccionamos 4 muestras de X_test
sample_X = X_test[sample_indices]
sample_Y_true = Y_test[sample_indices]

# Realizar predicciones
predictions_prob = modelo_iris.predict(sample_X)
predicted_classes = np.argmax(predictions_prob, axis=1)

# Convertir las etiquetas numéricas a nombres de especies
# Aseguramos que 'encoder' está disponible. Si no lo estuviera, tendríamos que re-instanciarlo y fittearlo.
# En este caso, ya está disponible en el entorno del notebook.
true_species = encoder.inverse_transform(sample_Y_true)
predicted_species = encoder.inverse_transform(predicted_classes)

print("\n--- Resultados de Predicción para Muestras Seleccionadas ---")
results = pd.DataFrame({
    'Características (SepalL, SepalW, PetalL, PetalW)': sample_X.tolist(),
    'Especie Real': true_species,
    'Especie Predicha': predicted_species,
    'Probabilidades': [np.round(p, 2).tolist() for p in predictions_prob]
})
display(results)
```

### Visualización de las Probabilidades de Predicción

```python
import matplotlib.pyplot as plt

# Seleccionamos la primera muestra de las predicciones para visualizar
# Puedes cambiar el índice para ver otras muestras (por ejemplo, sample_idx_to_plot = 1)
sample_idx_to_plot = 0
sample_pred_prob = predictions_prob[sample_idx_to_plot]
sample_true_species = true_species[sample_idx_to_plot]
sample_predicted_species = predicted_species[sample_idx_to_plot]

species_names = encoder.classes_ # Obtener los nombres originales de las clases

plt.figure(figsize=(8, 5))
plt.bar(species_names, sample_pred_prob, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.xlabel('Especie de Iris')
plt.ylabel('Probabilidad')
plt.title(f'Probabilidades de Predicción para una Muestra (Real: {sample_true_species}, Predicha: {sample_predicted_species})')
plt.ylim(0, 1)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
